## 1. DNA Sequence Input and Nucleotide Counting

DNA (Deoxyribonucleic Acid) contains four standard nitrogenous bases:

- Adenine (A)
- Thymine (T)
- Guanine (G)
- Cytosine (C)

In this section, Python is used to define a DNA sequence, calculate its length, and count the occurrence of each nucleotide.

These calculations provide the foundation for further sequence analysis.


```python

# DNA Sequence Analysis Toolkit
# Project 01

# Step 1: Define a DNA sequence
dna_sequence = "ATGGCTAACCGTTAG"

# Step 2: Calculate sequence length
sequence_length = len(dna_sequence)

# Step 3: Count nucleotides
count_A = dna_sequence.count("A")
count_T = dna_sequence.count("T")
count_G = dna_sequence.count("G")
count_C = dna_sequence.count("C")

# Step 4: Display the results
print("DNA Sequence:", dna_sequence)
print("Sequence Length:", sequence_length)

print("Adenine:", count_A)
print("Thymine:", count_T)
print("Guanine:", count_G)
print("Cytosine:", count_C)
```

## 2. GC and AT Content Analysis

DNA sequences contain four nucleotides: Adenine (A), Thymine (T), Guanine (G), and Cytosine (C).

- **GC content** represents the percentage of Guanine (G) and Cytosine (C) in a DNA sequence.
- **AT content** represents the percentage of Adenine (A) and Thymine (T).

GC and AT content are important characteristics in DNA analysis. They help describe the nucleotide composition of a sequence and can provide useful information about its properties.

In this section, Python is used to calculate both percentages and examine the nucleotide composition of the selected DNA sequence.



```python
# Calculate GC content
gc_count = dna_sequence.count("G") + dna_sequence.count("C")
gc_content = (gc_count / len(dna_sequence)) * 100

# Calculate AT content
at_count = dna_sequence.count("A") + dna_sequence.count("T")
at_content = (at_count / len(dna_sequence)) * 100

# Display results
print("GC Content:", round(gc_content, 2), "%")
print("AT Content:", round(at_content, 2), "%")
```


## 3. DNA Sequence Validation

Before analyzing a DNA sequence, it is important to ensure that the input contains only valid DNA nucleotides.

A standard DNA sequence consists of four bases:

- Adenine (A)
- Thymine (T)
- Guanine (G)
- Cytosine (C)

In this section, a Python function is created to validate DNA sequences by checking for invalid characters and empty inputs.

This helps prevent errors during subsequent sequence analysis.


```python
# DNA Sequence Validation
dna_sequence = "ATGGCTAACCGTTAG"

def validate_dna_sequence(sequence):

    # Convert the sequence to uppercase and remove extra spaces
    sequence = sequence.upper().strip()

    # Define the valid DNA nucleotides
    valid_bases = set("ATGC")

    # Identify invalid characters
    invalid_bases = set(sequence) - valid_bases

    # Check if the sequence is empty
    if not sequence:
        return False, sequence, "Error: DNA sequence is empty."

    # Check if invalid characters exist
    if invalid_bases:
        return False, sequence, f"Error: Invalid characters found: {invalid_bases}"

    return True, sequence, "DNA sequence is valid."


# Validate our DNA sequence
is_valid, dna_sequence, message = validate_dna_sequence(dna_sequence)

# Display the result
print(message)
print("Validated DNA Sequence:", dna_sequence)
```


## 4. DNA Sequence Analysis Using Biopython

Biopython is a Python library designed for biological computation and bioinformatics.

It provides specialized tools for working with biological sequences, including DNA, RNA, and proteins.

In this section, the validated DNA sequence is converted into a Biopython `Seq` object. This allows us to perform sequence analysis using built-in biological methods rather than relying only on basic Python operations.


```python
# Install Biopython
!pip install biopython
```

```python
# Import Biopython
from Bio.Seq import Seq

# Create a Biopython DNA sequence object
bio_dna = Seq(dna_sequence)

# Calculate GC content
gc_count = bio_dna.count("G") + bio_dna.count("C")
gc_content = (gc_count / len(bio_dna)) * 100

# Calculate AT content
at_count = bio_dna.count("A") + bio_dna.count("T")
at_content = (at_count / len(bio_dna)) * 100

# Display results
print("Biopython DNA Sequence:", bio_dna)
print("Sequence Length:", len(bio_dna))
print("GC Content:", round(gc_content, 2), "%")
print("AT Content:", round(at_content, 2), "%")
```


## 5. DNA Complement and Reverse Complement

DNA consists of two complementary strands that pair according to specific base-pairing rules:

- Adenine (A) pairs with Thymine (T).
- Guanine (G) pairs with Cytosine (C).

In this section, Biopython is used to generate:

- **Complementary sequence:** Replaces each nucleotide with its complementary base.
- **Reverse complement:** Generates the complementary sequence in reverse order.

These operations are fundamental to understanding DNA strand orientation and sequence analysis.


```python
# DNA Reverse Complement

# Generate the complementary DNA strand
complement_dna = bio_dna.complement()

# Generate the reverse complement
reverse_complement_dna = bio_dna.reverse_complement()

# Display results
print("Original DNA:", bio_dna)
print("Complement DNA:", complement_dna)
print("Reverse Complement:", reverse_complement_dna)
```


## 6. DNA Transcription into RNA

Transcription is the biological process in which a DNA sequence is used to produce an RNA molecule.

During transcription, Thymine (T) is replaced with Uracil (U) in the RNA sequence.

In this project, the DNA sequence is treated as the coding strand, so Biopython's `transcribe()` method converts it into the corresponding RNA sequence.

This step prepares the sequence for protein translation.


```python
# DNA Transcription into RNA

# Transcribe DNA into RNA
rna_sequence = bio_dna.transcribe()

# Display results
print("Original DNA:", bio_dna)
print("Transcribed RNA:", rna_sequence)
print("DNA Length:", len(bio_dna))
print("RNA Length:", len(rna_sequence))
```


## 7. RNA Translation into Protein

Translation is the biological process through which the information in an RNA sequence is used to produce a protein.

The RNA sequence is read in groups of three nucleotides called **codons**. Each codon corresponds to a specific amino acid or a stop signal.

In this section, Biopython's `translate()` method is used to convert the RNA sequence into its corresponding protein sequence.

The resulting protein sequence provides a simple demonstration of how genetic information is translated into amino acids.


```python
# RNA Translation into Protein

# Translate RNA into a protein sequence
protein_sequence = rna_sequence.translate()

# Display results
print("RNA Sequence:", rna_sequence)
print("Protein Sequence:", protein_sequence)
print("Protein Length:", len(protein_sequence))
```


## 8. Codon Analysis

A codon is a sequence of three nucleotides in RNA that specifies an amino acid or a stop signal during protein translation.

Since RNA is read in groups of three nucleotides, dividing the sequence into codons helps us understand how genetic information is organized.

In this section, Python is used to divide the RNA sequence into consecutive groups of three bases.

This provides a clearer view of the genetic code and prepares the sequence for further analysis, including open reading frame detection.


```python
# Codon Analysis

# Divide RNA sequence into groups of three nucleotides
codons = [
    str(rna_sequence[i:i+3])
    for i in range(0, len(rna_sequence), 3)
]

# Display codons
print("RNA Sequence:", rna_sequence)
print("Codons:", codons)
print("Total Codons:", len(codons))
```


## 9. Open Reading Frame (ORF) Analysis

An Open Reading Frame (ORF) is a sequence of nucleotides that begins with a start codon and ends with a stop codon.

In RNA:
- **Start codon:** AUG
- **Stop codons:** UAA, UAG, and UGA

ORF identification is an important step in bioinformatics because it helps locate potential protein-coding regions within a sequence.

In this project, Python is used to search for an ORF in the RNA sequence by identifying a start codon and the next in-frame stop codon.

**Note:** This implementation examines only one reading frame and is a simplified demonstration of ORF detection.


```python
# Open Reading Frame (ORF) Analysis

start_codon = "AUG"
stop_codons = {"UAA", "UAG", "UGA"}

# Find the ORF
orf_start = None
orf_end = None

for i in range(0, len(rna_sequence) - 2, 3):

    codon = str(rna_sequence[i:i+3])

    if codon == start_codon and orf_start is None:
        orf_start = i

    elif codon in stop_codons and orf_start is not None:
        orf_end = i + 3
        break

# Display results
if orf_start is not None and orf_end is not None:

    orf_sequence = rna_sequence[orf_start:orf_end]

    print("ORF Found!")
    print("ORF Sequence:", orf_sequence)
    print("ORF Length:", len(orf_sequence), "nucleotides")

else:
    print("No complete ORF found.")
```


## 10. Creating a FASTA File

FASTA is a widely used text-based format for storing biological sequences, including DNA, RNA, and protein sequences.

A FASTA file contains:
- A header line beginning with the `>` symbol, which identifies the sequence.
- One or more lines containing the biological sequence itself.

In this section, a sample FASTA file is created with two DNA sequences.

Using this format allows biological sequences to be stored, shared, and processed by bioinformatics tools.


```python
# Create a sample FASTA file

fasta_content = """>DNA_Sequence_1
ATGGCTAACCGTTAG
>DNA_Sequence_2
ATGCGTACCGGA
"""

with open("sample_sequences.fasta", "w") as fasta_file:
    fasta_file.write(fasta_content)

print("FASTA file created successfully!")
```


## 11. Reading and Analyzing FASTA Sequences

Bioinformatics projects often involve analyzing biological sequences stored in external files rather than manually entering them into Python.

Biopython provides the `SeqIO` module, which allows us to read and process sequence files in different biological formats.

In this section, the FASTA file created in the previous step is read using `SeqIO.parse()`.

For each sequence, the program extracts and analyzes:
- Sequence ID
- DNA sequence
- Sequence length
- Individual nucleotide counts
- GC content percentage

This demonstrates how multiple DNA sequences can be processed systematically using Biopython.


```python
# Read and Analyze FASTA File

from Bio import SeqIO

# Read all sequences from the FASTA file
for record in SeqIO.parse("sample_sequences.fasta", "fasta"):

    sequence = record.seq

    # Calculate nucleotide counts
    a_count = sequence.count("A")
    t_count = sequence.count("T")
    g_count = sequence.count("G")
    c_count = sequence.count("C")

    # Calculate GC content
    gc_content = ((g_count + c_count) / len(sequence)) * 100

    # Display sequence information
    print("\nSequence ID:", record.id)
    print("DNA Sequence:", sequence)
    print("Sequence Length:", len(sequence))
    print("A:", a_count)
    print("T:", t_count)
    print("G:", g_count)
    print("C:", c_count)
    print("GC Content:", round(gc_content, 2), "%")
```

```python
# Test GC and AT Content Calculations

def calculate_gc_at(sequence):

    if not sequence:
        raise ValueError("DNA sequence cannot be empty.")

    valid_bases = set("ATGC")

    if not set(sequence.upper()) <= valid_bases:
        raise ValueError("DNA sequence contains invalid characters.")

    sequence = sequence.upper()

    gc_count = sequence.count("G") + sequence.count("C")
    at_count = sequence.count("A") + sequence.count("T")

    gc_content = (gc_count / len(sequence)) * 100
    at_content = (at_count / len(sequence)) * 100

    return round(gc_content, 2), round(at_content, 2)


# Test the calculation
test_gc, test_at = calculate_gc_at("ATGGCTAACCGTTAG")

print("GC Content:", test_gc, "%")
print("AT Content:", test_at, "%")
print("Total:", round(test_gc + test_at, 2), "%")
```
